## 4. Data Cleaning and Transformation




### 4.1 Select Relevant Columns

* Before cleaning the dataset, we first examine missing values, duplicate records, data types, and unique values. This helps us understand what problems exist and decide how to handle them correctly.

```python
columns_to_use = [
    "crash_date",
    "crash_time",
    "borough",
    "number_of_persons_injured",
    "number_of_persons_killed",
    "number_of_pedestrians_injured",
    "number_of_pedestrians_killed",
    "number_of_cyclist_injured",
    "number_of_cyclist_killed",
    "number_of_motorist_injured",
    "number_of_motorist_killed",
    "contributing_factor_vehicle_1",
    "contributing_factor_vehicle_2",
    "vehicle_type_code1",
    "vehicle_type_code2",
    "collision_id"
]

df = df[columns_to_use].copy()
```

### 4.2 Missing Values

```python
df.isnull().sum()
```

```python
missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_data = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

missing_data
```

### 4.3 Duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate collision IDs:", df["collision_id"].duplicated().sum())
```

```python
df_clean = df.copy()
```

### 4.4 Data Types

```python
df_clean["crash_date"] = pd.to_datetime(
    df_clean["crash_date"],
    errors="coerce"
)
```

```python
df_clean["crash_hour"] = pd.to_numeric(
    df_clean["crash_time"].str.split(":").str[0],
    errors="coerce"
).astype("Int64")
```

```python
df_clean[
    ["crash_time", "crash_hour"]
].head()
```

```python
df_clean["crash_hour"].value_counts().sort_index()
```

### 4.5 Categorical Data Cleaning

```python
# Borough
df_clean["borough"] = (
    df_clean["borough"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Contributing factors
factor_columns = [
    "contributing_factor_vehicle_1",
    "contributing_factor_vehicle_2"
]

for column in factor_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
    )

# Vehicle types
vehicle_columns = [
    "vehicle_type_code1",
    "vehicle_type_code2"
]

for column in vehicle_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
    )
```

### 4.6 Injury and Fatality Validation

```python
# Total Persons Injured
# Total Persons Killed
# Pedestrians, Cyclists, and Motorists

calculated_injured = (
    df_clean["number_of_pedestrians_injured"]
    + df_clean["number_of_cyclist_injured"]
    + df_clean["number_of_motorist_injured"]
)

calculated_killed = (
    df_clean["number_of_pedestrians_killed"]
    + df_clean["number_of_cyclist_killed"]
    + df_clean["number_of_motorist_killed"]
)

df_clean["number_of_persons_injured"] = (
    df_clean["number_of_persons_injured"]
    .fillna(calculated_injured)
)

df_clean["number_of_persons_killed"] = (
    df_clean["number_of_persons_killed"]
    .fillna(calculated_killed)
)
```

```python
columns_to_drop = [
    "number_of_pedestrians_injured",
    "number_of_pedestrians_killed",
    "number_of_cyclist_injured",
    "number_of_cyclist_killed",
    "number_of_motorist_injured",
    "number_of_motorist_killed"
]

df_clean = df_clean.drop(columns=columns_to_drop)
```

```python
df_clean.columns
```

### 4.7 Create Analysis Variables

```python
df_clean["year"] = df_clean["crash_date"].dt.year
df_clean["month"] = df_clean["crash_date"].dt.month_name()
df_clean["day_of_week"] = df_clean["crash_date"].dt.day_name()

df_clean["crash_severity"] = "No Injury"

df_clean.loc[
    df_clean["number_of_persons_injured"] > 0,
    "crash_severity"
] = "Injury"

df_clean.loc[
    df_clean["number_of_persons_killed"] > 0,
    "crash_severity"
] = "Fatal"
```

## 5 Data analisis 🔢

### 5.1 Research Question 1

**Which NYC borough has the highest number of motor vehicle collisions?**

The number of collisions was grouped by borough to identify which area of New York City reported the highest number of motor vehicle crashes.

```python
# Count collisions by borough
borough_collisions = (
    df_clean["borough"]
    .value_counts()
    .sort_values(ascending=False)
)

borough_collisions
```

```python
# Count collisions by borough
borough_collisions = (
    df_clean["borough"]
    .dropna()
    .value_counts()
    .reset_index()
)

borough_collisions.columns = ["borough", "collisions"]

# Interactive bar chart
fig = px.bar(
    borough_collisions,
    x="borough",
    y="collisions",
    color="collisions",
    text="collisions",
    title="Motor Vehicle Collisions by Borough",
    labels={
        "borough": "Borough",
        "collisions": "Number of Collisions"
    }
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Borough",
    yaxis_title="Number of Collisions",
    showlegend=False
)

fig.show()
```

```python
top_borough = borough_collisions.iloc[0]["borough"]
top_collisions = borough_collisions.iloc[0]["collisions"]

print("Borough with the most collisions:", top_borough)
print("Number of collisions:", top_collisions)
```

**Interpretation:**  

Among collisions with a reported borough, **Brooklyn** had the highest number of motor vehicle collisions, with **506,806 recorded crashes**.

### 5.2 Research Question 2

**At what times and on which days do motor vehicle collisions occur most frequently?**

The number of collisions was analyzed by hour of the day and day of the week to identify when crashes occur most frequently.

```python
# Collisions by hour
hour_collisions = (
    df_clean["crash_hour"]
    .value_counts()
    .sort_index()
)

hour_collisions

```

```python
# Collisions by hour
hour_collisions = (
    df_clean["crash_hour"]
    .dropna()
    .value_counts()
    .sort_index()
    .reset_index()
)

hour_collisions.columns = ["crash_hour", "collisions"]

# Interactive line chart
fig = px.line(
    hour_collisions,
    x="crash_hour",
    y="collisions",
    markers=True,
    title="Motor Vehicle Collisions by Hour of Day",
    labels={
        "crash_hour": "Hour of Day",
        "collisions": "Number of Collisions"
    }
)

fig.update_layout(
    xaxis=dict(tickmode="linear", dtick=1),
    xaxis_title="Hour of Day",
    yaxis_title="Number of Collisions"
)

fig.show()
```

```python
day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

day_collisions = (
    df_clean["day_of_week"]
    .value_counts()
    .reindex(day_order)
    .reset_index()
)

day_collisions.columns = ["day_of_week", "collisions"]
```

```python
# Interactive bar chart
fig = px.bar(
    day_collisions,
    x="day_of_week",
    y="collisions",
    color="collisions",
    text="collisions",
    title="Motor Vehicle Collisions by Day of Week",
    labels={
        "day_of_week": "Day of Week",
        "collisions": "Number of Collisions"
    }
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Day of Week",
    yaxis_title="Number of Collisions",
    showlegend=False
)

fig.show()
```

```python
print("Hour with most collisions:", hour_collisions.idxmax())
print("Day with most collisions:", day_collisions.idxmax())
```

**Interpretation:**  
Motor vehicle collisions occurred most frequently around **4:00 PM (16:00)**, while **Friday** had the highest number of recorded crashes.

### 5.3 Research Question 3

**Which contributing factors and vehicle types are most common among injury or fatal crashes?**

This analysis focuses only on crashes classified as `"Injury"` or `"Fatal"` to identify the most frequently recorded contributing factors and vehicle types.

```python
# Keep only injury or fatal crashes
severe_crashes = df_clean[
    df_clean["crash_severity"].isin(["Injury", "Fatal"])
]

# Exclude "Unspecified"
top_factors = (
    severe_crashes[
        severe_crashes["contributing_factor_vehicle_1"] != "Unspecified"
    ]["contributing_factor_vehicle_1"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_factors.columns = ["factor", "collisions"]
```

```python
# Interactive horizontal bar chart
fig = px.bar(
    top_factors.sort_values("collisions"),
    x="collisions",
    y="factor",
    orientation="h",
    color="collisions",
    text="collisions",
    title="Top Contributing Factors in Injury or Fatal Crashes",
    labels={
        "factor": "Contributing Factor",
        "collisions": "Number of Crashes"
    }
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Number of Crashes",
    yaxis_title="Contributing Factor",
    showlegend=False
)

fig.show()
```

```python
#Top vehicles
top_vehicles = (
    severe_crashes["vehicle_type_code1"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_vehicles.columns = ["vehicle_type", "collisions"]
```

```python
# Interactive horizontal bar chart
fig = px.bar(
    top_vehicles.sort_values("collisions"),
    x="collisions",
    y="vehicle_type",
    orientation="h",
    color="collisions",
    text="collisions",
    title="Top Vehicle Types in Injury or Fatal Crashes",
    labels={
        "vehicle_type": "Vehicle Type",
        "collisions": "Number of Crashes"
    }
)

fig.update_traces(textposition="outside")
fig.update_layout(
    xaxis_title="Number of Crashes",
    yaxis_title="Vehicle Type",
    showlegend=False
)

fig.show()
```

```python
print(
    "Most common contributing factor:",
    top_factors.iloc[0]["factor"]
)

print(
    "Most common vehicle type:",
    top_vehicles.iloc[0]["vehicle_type"]
)
```

**Interpretation:**  
The analysis shows that **Driver Inattention/Distraction** is the most common contributing factor among injury or fatal crashes, while **Sedan** is the most frequently recorded vehicle type in these collisions.

3. **Driver Inattention/Distraction** was the most frequently recorded contributing factor among injury or fatal crashes, although it does not identify a specific cause. **Sedan** was the most frequently recorded vehicle type in these crashes.

The analysis of NYC motor vehicle collisions revealed clear patterns related to location, time, contributing factors, and vehicle types.
